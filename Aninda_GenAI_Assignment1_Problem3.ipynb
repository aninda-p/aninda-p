```python
# 3. Problem Statement

# Write a Python program to draw a 3D plot that visualizes the regression model for house price prediction using suitable Python-based 3D plotting libraries.
# Assume the following features were used:
# Area (sq ft)
# Number of Bedrooms
# House Price

import pandas as pd

data = {
    "Area(sqft)": [900, 1400, 2200, 2000, 2500, 3000],
    "Num of Bedrooms": [2, 3, 4, 3, 4, 5],
    "House Price": [180000, 260000, 330000, 280000, 350000, 430000]
}

df = pd.DataFrame(data)
df

# splitting dataset into features (X- things the model uses to make predictions) and the target variable (Y- what we want the model to predict).
# "House Price" = f("Area(sqft)", "Num of Bedrooms")
X = df[["Area(sqft)", "Num of Bedrooms"]]
Y = df["House Price"]

# create a regression model
from sklearn.linear_model import LinearRegression
reg_model = LinearRegression() # creates empty model
reg_model.fit(X, Y) # model analyzes the data and learns

# draw a 3D plot
import matplotlib.pyplot as plt
plt.rcParams["figure.figsize"] = (7, 3.5) # sets the default size of all plots
plt.rcParams["figure.autolayout"] = True # auto adjust layout

fig = plt.figure() # blank canvas to plot
ax = fig.add_subplot(111, projection="3d") # add 3d subplot to the canvas, 111: 1 row, 1 col, 1 subplot

# set x, y z labels
ax.set_xlabel("Area(sqft)")
ax.set_ylabel("Bedrooms")
ax.set_zlabel("House Price")
ax.set_title("3D plot that visualizes the regression model for house price prediction")
# scatter actual data
ax.scatter(df["Area(sqft)"], df["Num of Bedrooms"], df["House Price"], c=df["House Price"]) # c=df["House Price"], color mapping
plt.show()
```
